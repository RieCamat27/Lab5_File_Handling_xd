```python
from pathlib import Path
from datetime import datetime 
import shutil
import stat 

# Student Info
student_id = "2025-0706"
student_name = "Angel Marie P. Camat"

# Document Path 
documents_path = Path.home() / "Documents" / "Activity_5_Exercises"
documents_path.mkdir(parents=True, exist_ok=True)
file_to_copy = documents_path / f"intro_{student_id}.txt"
backup_file = documents_path / f"backup_intro_{student_id}.txt"
backup_file.mkdir(parents=True, exist_ok=True)

# Timestamp 
timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")

# Backup File Name
backup_file_name = f"backup_intro_{student_id}_{timestamp}.txt"
backup_file_path = Path.home() / "Documents" / "Activity_5_Exercises" / f"backup_{student_id}" / backup_file_name

# Copy File to Backup Location
# Create backup directory if it doesn't exist
backup_file_path.parent.mkdir(parents=True, exist_ok=True)

# Copy file to backup location
shutil.copy(file_to_copy, backup_file_path)

# Log File 
log_file = Path.home() / "Documents" / "Activity_5_Exercises" / f"backup_{student_id}/" 

# Fix: Remove the incorrect mkdir on backup_file (it's a file, not a directory)
# The directory creation is already handled by backup_file_path.parent.mkdir()

# Fix: Set log_file to point to a log file, not a directory
log_file = Path.home() / "Documents" / "Activity_5_Exercises" / f"backup_{student_id}" / "backup_log.txt"

# Write to log file
with log_file.open("a") as log:
    log.write(f"{datetime.now()}: Backed up {file_to_copy} to {backup_file_path}\n")
    log.write(f"{student_id}: {student_name}\n")
    log.write(f"Size: {backup_file_path.stat().st_size} bytes\n")
    

# Displayed Confirmation Message

print(f"Backup completed for {student_id}  ({student_name})")
print(f"File saved as: {backup_file_name}")
print(f"Log updated at: {log_file}")




```
